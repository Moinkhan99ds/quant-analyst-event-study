```python
def classify(row):

    subcat = str(row["SUBCATNAME"]).lower()
    text = str(row["text"]).lower()

    if "financial results" in subcat:
        return "Financial Results"

    elif ("board meeting" in subcat or
          "change in management" in subcat or
          "change in director" in subcat or
          "appointment" in subcat or
          "cessation" in subcat or
          "retirement" in subcat or
          "resignation" in subcat):
        return "Board / Management"

    elif ("investor" in subcat or
          "analyst" in subcat or
          "earnings call" in subcat):
        return "Investor Communication"

    elif ("agm" in subcat or
          "egm" in subcat or
          "allotment" in subcat or
          "esop" in subcat or
          "postal ballot" in subcat or
          "record date" in subcat or
          "dividend" in subcat):
        return "Corporate Actions"

    elif ("order" in subcat or
          "acquisition" in subcat or
          "agreement" in subcat or
          "mou" in subcat or
          "diversification" in subcat or
          "disinvestment" in subcat):
        return "Orders / Business"

    elif ("reg." in subcat or
          "reg-" in subcat or
          "compliance" in subcat or
          "certificate" in subcat or
          "clarification" in subcat):
        return "Regulatory / Compliance"

    else:
        return "Press / Public Updates"
```

```python
def get_returns(row):

    company = row["company"]
    event_date = row["date"].date()

    dates = market[market["company"] == company]["date"].dt.date.unique()

    future_dates = []

    for d in dates:
        if d > event_date:
            future_dates.append(d)

    if len(future_dates) < 20:
        return None

    d_close = market[
        (market["company"] == company) &
        (market["date"].dt.date == event_date)
    ]["close"].iloc[-1]

    d1 = market[
        (market["company"] == company) &
        (market["date"].dt.date == future_dates[0])
    ]["close"].iloc[-1]

    d5 = market[
        (market["company"] == company) &
        (market["date"].dt.date == future_dates[4])
    ]["close"].iloc[-1]

    d10 = market[
        (market["company"] == company) &
        (market["date"].dt.date == future_dates[9])
    ]["close"].iloc[-1]

    d20 = market[
        (market["company"] == company) &
        (market["date"].dt.date == future_dates[19])
    ]["close"].iloc[-1]

    return [
        (d1 - d_close) / d_close * 100,
        (d5 - d_close) / d_close * 100,
        (d10 - d_close) / d_close * 100,
        (d20 - d_close) / d_close * 100
    ]
```
